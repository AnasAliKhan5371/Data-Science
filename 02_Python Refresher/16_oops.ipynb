```python
class employee:
    company="HP" #class attributes
    def __init__(self,name,salary): #instance attributes
        self.name=name
        self.salary=salary
    def printDetails(self):
        print(f"company of {self.name} is {self.company} with salary Rs{self.salary}")

e=employee("harry",35000)
e.printDetails()
```

```python
class employee:
    company="HP" #class attributes
    def __init__(self,name,salary): #instance attributes
        self.name=name
        self.salary=salary
    def printDetails(self):
        print(f"company of {self.name} is {self.company} with salary Rs{self.salary}")
    @staticmethod
    def printTime():
        print("time is now")
        
e=employee("harry",35000)
e.printTime()
```

```python
class employee:
    company="HP" #class attributes
    def __init__(self,name,salary): #instance attributes
        self.name=name
        self.salary=salary
    def printDetails(self):
        print(f"company of {self.name} is {self.company} with salary Rs{self.salary}")
    @classmethod
    def printClassDetails(cls):
        print(f"company is {cls.company}")
        
e=employee("harry",35000)
e.printClassDetails()
```
