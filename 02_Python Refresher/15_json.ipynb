```python
import json
```

```python
#json.dumps() – Convert Python object to JSON string
obj={"name":"anas","company":"google","salary":25000}
json.dumps(obj) # dump as a string
```

```python
#json.loads() – Convert JSON string to Python object
d=json.loads('{"name": "anas", "company": "google", "salary": 25000}')
```

```python
type(d)
```

```python
# json.dump() – Write JSON data to a file
f=open("data.json","w")
json.dump(obj,f)
f.close()
```

```python
#json.load() – Read JSON data from a file
with open("data.json", "r") as f:
    data = json.load(f)
print(data)
```
