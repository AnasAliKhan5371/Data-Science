```python
import json
```

```python
# func. to load data
def load_data(filename):
    with open(filename,"r") as f:
        data=json.load(f)
    return data
```

```python
data=load_data("data.json")
```

```python
# func. to display users & connections
def display_users(data):
    print("Users and their connections\n")
    for user in data['users']:
        print(f"ID:{user["id"]} - {user['name']} is friends with {user['friends']} and liked pager are {user['liked_pages']}")
    print("\nPages information")
    for page in data['pages']:
        print(f"{page['id']}: {page['name']}")        

display_users(data)
```
