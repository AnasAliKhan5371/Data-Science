```python
import pandas as pd
employees = pd.DataFrame({
    "EmpID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "DeptID": [10, 20, 30]
})

departments = pd.DataFrame({
    "DeptID": [10, 20, 40],
    "DeptName": ["HR", "Engineering", "Marketing"]
})
```

```python
employees
```

```python
departments
```

```python
#Inner Join (default)
pd.merge(employees,departments,on="DeptID") 
```

```python
#Left Join
pd.merge(employees, departments, on="DeptID", how="left")
```

```python
#Right Join
pd.merge(employees, departments, on="DeptID", how="right")
```

```python
#Outer Join
pd.merge(employees, departments, on="DeptID", how="outer")
```

```python
#Concatenating DataFrames
#Vertical (rows)
df1 = pd.DataFrame({"Name": ["Alice", "Bob"]})
df2 = pd.DataFrame({"Name": ["Charlie", "David"]})

pd.concat([df1, df2])
```

```python
df1 = pd.DataFrame({"Name": ["Alice", "Bob"],"Score":[443,221]})
df2 = pd.DataFrame({"Name": ["Charlie", "David"],"age":[33,44]})

pd.concat([df1, df2])
```

```python
#Horizontal (columns)
df1 = pd.DataFrame({"ID": [1, 2]})
df2 = pd.DataFrame({"Score": [90, 80]})

pd.concat([df1, df2], axis=1)
```
