```python
import pandas as pd
```

```python
#Series → One-dimensional (like a single column in Excel)
s1=pd.Series([71,84,55,77,98,46,66])
type(s1)
```

```python
s1 #auto-indexing
```

```python
#custom indexing
s2=pd.Series([71,84,55,77,98,46,66],index=["anas","ahmad","rohan","akash","harry","kirti","pooja"])
s2
```

```python
print(s2["harry"])
```

```python
# DataFrame → Two-D(like a full spreadsheet or SQL table)

data={
    "name":["alice","bob","charlie"],
    "age":[25,20,30],
    "city":["delhi","mumbai","bangalore"]
}
```

```python
df=pd.DataFrame(data)
df
```
