```python
import pandas as pd
df=pd.read_csv("data_cleaning_sample.csv")
df
```

```python
# check for null-true for null
df.isnull()
```

```python
# Count missing per column
df.isnull().sum()
```

```python
# Drop Missing Data
# Drop rows with *any* missing values
df.dropna()
```

```python
# Drop columns with missing values
df.dropna(axis=1)
```

```python
# Fill Missing Data
# # Replace NaN with 0
df.fillna(0)
```

```python
# Replace with mean
df["Age"].fillna(df["Age"].mean())
```

```python
df.ffill()      # Forward fill
```

```python
df.bfill()      # Backward fill
```

```python
# Detecting & Removing Duplicates
# True for duplicates
df.duplicated()
```

```python
# Remove duplicate rows
df.drop_duplicates()     
```

```python
# Check based on specific columns
df.duplicated(subset=["Name", "Age"])
```

```python
#String Operations
# Converts all names to lowercase.
df["Name"].str.lower() 
```

```python
# Checks if 'delhi' is in the city name, case-insensitive.
df["City"].str.contains("delhi", case=False)
```

```python
# Outputs a pandas Series where each element is a list of strings (the split parts).
#This is where a Python list comes into play, but the outer object is still a pandas Series.
df["Email"].str.split("@")
```

```python
#Type Conversions
df2=df.dropna().copy()
df2["Age"] = df2["Age"].astype(int)
df2.info()
```

```python
# applying functions
#.apply()-Apply any function to rows or columns
df2["Age Group"] = df2["Age"].apply(lambda x: "Adult" if x >= 25 else "Minor")
df2
```

```python
def isminor(x):
    return "Adult" if x >= 25 else "Minor"
df2["Age Group"] = df2["Age"].apply(isminor)
df2
```

```python
#.map() → Element-wise mapping for Series
gender_map = {"M": "Male", "F": "Female"}
df2["Gender"] = df2["Gender"].map(gender_map)
df2
```

```python
#.replace() → Replace specific values
df["City"]=df["City"].replace({"Delhi": "New Delhi", "Mumbai": "New Mumbai"})
df
```
